```python
#Создаем пивот по пользователями с необходимыми метриками + оставляем order_dt и order_ts для присоединения доп. аттрибутов в будущем
df_users_pivot = pd.pivot_table(
    df,
    index='user_id',
    values=['order_dt', 'order_ts', 'order_id', 'revenue_rub', 'tickets_count', 'days_since_prev'],
    aggfunc={
        'order_dt': ['min', 'max'],
        'order_ts': ['min', 'max'],
        'order_id': 'nunique',
        'revenue_rub': 'mean',
        'tickets_count': 'mean',
        'days_since_prev': 'mean'
    }
)

#Переименовываем метрики
df_users_pivot.columns = [
    'days_prev_mean',
    'order_dt_max',
    'order_dt_min',
    'unique_orders_count',
    'order_ts_max', 
    'order_ts_min',
    'revenue_rub_mean',
    'tickets_count_mean'
]
```

```python
#Присоединеяем доп. атрибуты
df_users_pivot_info = df_users_pivot.merge(df, how='left', left_on=['user_id','order_ts_min'], right_on=['user_id','order_ts'])
```

```python
#Добавляем два булевых атрибута
df_users_pivot_info['is_two'] = (df_users_pivot_info['unique_orders_count'] >= 2).astype(bool)
df_users_pivot_info['is_five'] = (df_users_pivot_info['unique_orders_count'] >= 5).astype(bool)
```

```python
#Формируем список колонок для корректировки из местоположения в df
new_column_order = [
    'user_id', 
    'order_dt_min',
    'order_dt_max',
    'device_type_canonical',
    'region_name',
    'service_name',
    'event_type_main',
    'unique_orders_count',
    'revenue_rub_mean',
    'tickets_count_mean',
    'days_prev_mean',
    'is_two',
    'is_five'
]
```

```python
#Применяем список колонок
df_users_pivot_info = df_users_pivot_info[new_column_order]
```

```python
#Переименовываем колонки для читабельности
df_users_pivot_info = df_users_pivot_info.rename(columns={'device_type_canonical': 'first_device_type_canonical',
                                    'region_name': 'first_region_name',
                                    'service_name': 'first_service_name',
                                    'event_type_main': 'first_event_type_main'})
```

**Промежуточный вывод:**
- Создали пивот по пользователям `df_users_pivot_info`, где нашли и посчитали:
    - дату первого и последнего заказа;

    - Билетный оператор: Наибольшая лояльность наблюдается у операторов "Быстрый кассир" (85%), "Билет по телефону" (83%), но при небольшой клиентской базе.

- Влияние выручки и состава заказа

    - Выручка: Пользователи с повторными заказами концентрируются в диапазоне средней выручки 400-600 рублей, в то время как одноразовые покупатели - в диапазоне 0-100 рублей.

    - Количество билетов: Наиболее лояльны пользователи, покупающие 2-3 билета в среднем за заказ (74% возвратов). Пользователи с покупками 5+ билетов демонстрируют наименьшую лояльность (19%).

- Временные характеристики

    - День недели: Влияет на вероятность возврата минимально (разница между лучшим и худшим днем всего 3.6 процентных пункта).

    - Интервалы между покупками: Постоянные клиенты (5+ заказов) покупают в 2 раза чаще (средний интервал 9.9 дней), чем пользователи с 2-4 заказами (21.3 дня).

- Корреляционный анализ. Наиболее сильную связь с количеством заказов демонстрируют:

    - Частота покупок (0.39) - меньшие интервалы между заказами связаны с большим общим числом заказов

    - Количество билетов в заказе (0.38) - покупатели большего количества билетов склонны делать больше заказов

    - Средняя выручка (0.33) - более высокая средняя выручка коррелирует с большим количеством заказов
