Для выгрузки используйте запрос из предыдущего урока и библиотеку SQLAlchemy.

Выгрузка из базы данных SQL должна позволить собрать следующие данные:

- `user_id` — уникальный идентификатор пользователя, совершившего заказ;

````python
query = '''
select 
user_id,
device_type_canonical,
order_id,
created_dt_msk as order_dt,
created_ts_msk as order_ts,
currency_code,
revenue,
tickets_count,
extract(epoch from (created_dt_msk - lag(created_dt_msk) over (partition by user_id ORDER BY created_dt_msk)))/86400 as days_since_prev,
ps.event_id,
 event_name_code as event_name,
 event_type_main,
 service_name,
 region_name,
 city_name
from afisha.purchases as ps 
join afisha.events as es on es.event_id = ps.event_id
left join afisha.city as cs on cs.city_id = es.city_id 
left join afisha.regions as rs on rs.region_id = cs.region_id
where device_type_canonical IN ('mobile', 'desktop') AND 
es.event_type_main NOT IN ('фильм')
order by user_id
'''
````

### 3. Создание профиля пользователя

В будущем отдел маркетинга планирует создать модель для прогнозирования возврата пользователей. Поэтому сейчас они просят вас построить агрегированные признаки, описывающие поведение и профиль каждого пользователя.

**Задача 3.1.** Постройте профиль пользователя — для каждого пользователя найдите:

- дату первого и последнего заказа;
- устройство, с которого был сделан первый заказ;
- регион, в котором был сделан первый заказ;
- билетного партнёра, к которому обращались при первом заказе;
- жанр первого посещённого мероприятия (используйте поле `event_type_main`);
- общее количество заказов;
- средняя выручка с одного заказа в рублях;
- среднее количество билетов в заказе;
- среднее время между заказами.

После этого добавьте два бинарных признака:

- `is_two` — совершил ли пользователь 2 и более заказа;
- `is_five` — совершил ли пользователь 5 и более заказов.

**Рекомендация:** перед тем как строить профиль, отсортируйте данные по времени совершения заказа.

---


````python
# Сначала сортируем данные по времени заказа
df_sorted = df.sort_values(['user_id', 'order_ts'])
````

````python
# Создаем агрегированные метрики для пользователей
df_users_agg = df_sorted.groupby('user_id').agg(
    # Дата и время первого и последнего заказа
    first_order_dt=('order_dt', 'first'),
    last_order_dt=('order_dt', 'last'),
    first_order_ts=('order_ts', 'first'),
    last_order_ts=('order_ts', 'last'),
    
    # Категориальные признаки первого заказа (используем 'first' для получения первого значения)
    first_device_type=('device_type_canonical', 'first'),
    first_region=('region_name', 'first'),
    first_service=('service_name', 'first'),
    first_event_type=('event_type_main', 'first'),
    
    # Количественные метрики
    total_orders=('order_id', 'nunique'),
    avg_revenue=('revenue_rub', 'mean'),
    avg_tickets=('tickets_count', 'mean'),
    avg_days_between_orders=('days_since_prev', 'mean')
).reset_index()
````

````python
# Вычисляем среднее время между заказами (в днях)
# Для пользователей с одним заказом days_since_prev будет NaN, что корректно
# Можно заменить на 0 или оставить как NaN
df_users_agg['avg_days_between_orders'] = df_users_agg['avg_days_between_orders']

# Добавляем бинарные признаки
df_users_agg['is_two'] = (df_users_agg['total_orders'] >= 2).astype(bool)
df_users_agg['is_five'] = (df_users_agg['total_orders'] >= 5).astype(bool)
````

````python
# Переименовываем колонки для единообразия
df_users_agg = df_users_agg.rename(columns={
    'first_order_dt': 'order_dt_min',
    'last_order_dt': 'order_dt_max',
    'avg_revenue': 'revenue_rub_mean',
    'avg_tickets': 'tickets_count_mean',
    'avg_days_between_orders': 'days_prev_mean',
    'first_device_type': 'first_device_type_canonical',
    'first_region': 'first_region_name',
    'first_service': 'first_service_name',
    'first_event_type': 'first_event_type_main',
    'total_orders': 'unique_orders_count'
})

# Приводим к нужному порядку колонок
new_column_order = [
    'user_id', 
    'order_dt_min',
    'order_dt_max',
    'first_device_type_canonical',
    'first_region_name',
    'first_service_name',
    'first_event_type_main',
    'unique_orders_count',
    'revenue_rub_mean',
    'tickets_count_mean',
    'days_prev_mean',
    'is_two',
    'is_five'
]

#Применяем список колонок
df_users_pivot_info = df_users_agg[new_column_order]
````
